# to compare two variables we use joint plot

```python
iris.head(5)
```

```python
plt.figure(figsize=(8,6))
sns.jointplot(data= iris, x= iris['PetalLengthCm'],y= iris['PetalWidthCm'],hue= iris['Species'],kind= 'scatter')
```

```python
plt.figure(figsize=(8,6))
sns.jointplot(data= iris, x= iris['SepalLengthCm'],y= iris['SepalWidthCm'],kind= 'reg')
```

```python
plt.figure(figsize=(8,6))
sns.boxplot(iris,x= iris['Species'],y= iris['PetalLengthCm'])
```

```python
iris.boxplot(by= 'Species',figsize=(12,7))
```

# combine 2 plots in one

```python
fig= sns.boxplot(iris,x= 'Species',y= iris['SepalLengthCm'])
fig= sns.stripplot(iris,x= 'Species',y= iris['SepalLengthCm'],jitter= True,edgecolor='grey')
```

```python
plt.figure(figsize=(15,10))
plt.subplot(2,2,1)
sns.violinplot(x='Species',y='PetalLengthCm',data=iris,color= 'red',hue= 'Species')
plt.subplot(2,2,2)
sns.violinplot(x='Species',y='PetalWidthCm',data=iris,color= 'blue',hue= 'Species')
plt.subplot(2,2,3)
sns.violinplot(x='Species',y='SepalLengthCm',data=iris,color= 'green',hue= 'Species')
plt.subplot(2,2,4)
sns.violinplot(x='Species',y='SepalWidthCm',data=iris,color= 'pink',hue= 'Species')
```
